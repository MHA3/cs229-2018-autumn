### **3(a) [3 points] Score function**

Let the distribution of a random variable $Y$ parameterized by $\theta \in \mathbb{R}^n$ be $p(y; \theta)$.

The score function associated with $p(y; \theta)$ is defined as $\nabla_\theta \log p(y; \theta)$, which signifies the sensitivity of the likelihood function with respect to the parameters. Note that the score function is actually a vector since it’s the gradient of a scalar quantity with respect to the vector $\theta$.

Recall that $\mathbb{E}_{y\sim p(y)}[g(y)] = \int_{-\infty}^{\infty} p(y)g(y)dy$. Using this fact, show that the expected value of the score is 0, i.e.

$$ \mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta'} \log p(y; \theta')|_{\theta'=\theta}] = 0 $$

Would you like the text for problem **3(b)** on Fisher Information next?

### Solution 3(a)

We want to show that $\mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta)] = 0$.

By definition of expectation for a continuous random variable:
$$
\mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta)] = \int_{-\infty}^{\infty} p(y; \theta) \nabla_{\theta} \log p(y; \theta) dy
$$

Recall the derivative of a logarithm: $\nabla_\theta \log f(\theta) = \frac{\nabla_\theta f(\theta)}{f(\theta)}$. Applying this to the log-likelihood:
$$
\nabla_{\theta} \log p(y; \theta) = \frac{\nabla_{\theta} p(y; \theta)}{p(y; \theta)}
$$

Substitute this back into the expectation integral:
$$
\begin{aligned}
\mathbb{E}[\nabla_{\theta} \log p(y; \theta)] &= \int_{-\infty}^{\infty} p(y; \theta) \frac{\nabla_{\theta} p(y; \theta)}{p(y; \theta)} dy \\
&= \int_{-\infty}^{\infty} \nabla_{\theta} p(y; \theta) dy
\end{aligned}
$$

Assuming regularity conditions that allow us to interchange the order of differentiation and integration (Leibniz integral rule):
$$
\int_{-\infty}^{\infty} \nabla_{\theta} p(y; \theta) dy = \nabla_{\theta} \left( \int_{-\infty}^{\infty} p(y; \theta) dy \right)
$$

Since $p(y; \theta)$ is a valid probability density function, it must integrate to 1 over its domain:
$$
\int_{-\infty}^{\infty} p(y; \theta) dy = 1
$$

Therefore:
$$
\nabla_{\theta} (1) = 0
$$

Thus, we have shown that:
$$
\mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta)] = 0
$$

### **3(b) [2 points] Fisher Information**

Let us now introduce a quantity known as the Fisher information. It is defined as the covariance matrix of the score function,
$$I(\theta) = \text{Cov}_{y\sim p(y;\theta)}[\nabla_{\theta'} \log p(y; \theta')|_{\theta'=\theta}]$$

Intuitively, the Fisher information represents the amount of information that a random variable $Y$ carries about a parameter $\theta$ of interest. When the parameter of interest is a vector (as in our case, since $\theta \in \mathbb{R}^n$), this information becomes a matrix. Show that the Fisher information can equivalently be given by
$$I(\theta) = E_{y\sim p(y;\theta)}[\nabla_{\theta'} \log p(y; \theta') \nabla_{\theta'} \log p(y; \theta')^T |_{\theta'=\theta}]$$

Note that the Fisher Information is a function of the parameter. The parameter of the Fisher information is both a) the parameter value at which the score function is evaluated, and b) the parameter of the distribution with respect to which the expectation and variance is calculated.

### Solution 3(b)

We are given the definition of Fisher Information as the covariance of the score function:
$$
I(\theta) = \text{Cov}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta)]
$$

Recall the definition of the covariance matrix for a random vector $Z$:
$$
\text{Cov}[Z] = \mathbb{E}[ZZ^T] - \mathbb{E}[Z]\mathbb{E}[Z]^T
$$

Let $Z = \nabla_{\theta} \log p(y; \theta)$. From part 3(a), we proved that the expected value of the score function is zero:
$$
\mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta)] = 0
$$

Substituting $\mathbb{E}[Z] = 0$ into the covariance formula:
$$
\begin{aligned}
I(\theta) &= \mathbb{E}[ZZ^T] - 0 \cdot 0^T \\
&= \mathbb{E}[ZZ^T] \\
&= \mathbb{E}_{y\sim p(y;\theta)}[\nabla_{\theta} \log p(y; \theta) \nabla_{\theta} \log p(y; \theta)^T]
\end{aligned}
$$

Thus, we have shown the alternate form of the Fisher Information.

### **3(c) [5 points] Fisher Information (alternate form)**

It turns out that the Fisher Information can not only be defined as the covariance of the score function, but in most situations it can also be represented as the expected negative Hessian of the log-likelihood.

Show that:
$$ \mathbb{E}_{y\sim p(y;\theta)}[-\nabla^2_{\theta'} \log p(y; \theta')|_{\theta'=\theta}] = I(\theta) $$

**Remark.** The Hessian represents the curvature of a function at a point. This shows that the expected curvature of the log-likelihood function is also equal to the Fisher information matrix. If the curvature of the log-likelihood at a parameter is very steep (i.e, Fisher Information is very high), this generally means you need fewer number of data samples to a estimate that parameter well (assuming data was generated from the distribution with those parameters), and vice versa. The Fisher information matrix associated with a statistical model parameterized by $\theta$ is extremely important in determining how a model behaves as a function of the number of training set examples.

### Solution 3(c)

We start with the identity that the probability density function integrates to 1:
$$ \int p(y; \theta) dy = 1 $$

Taking the gradient with respect to $\theta$ (as done in 3(a)):
$$ \nabla_\theta \int p(y; \theta) dy = \int \nabla_\theta p(y; \theta) dy = 0 $$

Now, we take the derivative (gradient) a second time with respect to $\theta$. Since we are dealing with vectors, this results in a Hessian matrix.
$$ \nabla_\theta \left( \int \nabla_\theta p(y; \theta) dy \right) = \int \nabla_\theta^2 p(y; \theta) dy = 0 $$
(Here $\nabla^2$ denotes the Hessian matrix).

We need to express $\nabla_\theta^2 p(y; \theta)$ in terms of log-likelihoods.
Recall $\nabla_\theta p(y; \theta) = p(y; \theta) \nabla_\theta \log p(y; \theta)$.

Differentiate this product with respect to $\theta$:
$$
\begin{aligned}
\nabla_\theta^2 p(y; \theta) &= \nabla_\theta \left( p(y; \theta) \nabla_\theta \log p(y; \theta) \right) \\
&= \nabla_\theta p(y; \theta) (\nabla_\theta \log p(y; \theta))^T + p(y; \theta) \nabla_\theta^2 \log p(y; \theta)
\end{aligned}
$$

Substitute $\nabla_\theta p(y; \theta) = p(y; \theta) \nabla_\theta \log p(y; \theta)$ again into the first term:
$$
\nabla_\theta^2 p(y; \theta) = p(y; \theta) (\nabla_\theta \log p(y; \theta)) (\nabla_\theta \log p(y; \theta))^T + p(y; \theta) \nabla_\theta^2 \log p(y; \theta)
$$

Now substitute this back into the integral equation $\int \nabla_\theta^2 p(y; \theta) dy = 0$:
$$
\int \left[ p(y; \theta) (\nabla_\theta \log p(y; \theta)) (\nabla_\theta \log p(y; \theta))^T + p(y; \theta) \nabla_\theta^2 \log p(y; \theta) \right] dy = 0
$$

By linearity of the integral (expectation):
$$
\int p(y; \theta) (\nabla_\theta \log p(y; \theta)) (\nabla_\theta \log p(y; \theta))^T dy + \int p(y; \theta) \nabla_\theta^2 \log p(y; \theta) dy = 0
$$

Recognizing these integrals as expectations:
$$
\mathbb{E}[(\nabla_\theta \log p(y; \theta)) (\nabla_\theta \log p(y; \theta))^T] + \mathbb{E}[\nabla_\theta^2 \log p(y; \theta)] = 0
$$

The first term is exactly the definition of Fisher Information $I(\theta)$ derived in 3(b) (since expected score is 0).
$$
I(\theta) + \mathbb{E}[\nabla_\theta^2 \log p(y; \theta)] = 0
$$

Rearranging gives the desired result:
$$
I(\theta) = - \mathbb{E}[\nabla_\theta^2 \log p(y; \theta)]
$$

### 3(d) [5 points] Approximating $D_{KL}$ with Fisher Information

As we explained at the start of this problem, we are interested in the set of all distributions that are at a small fixed $D_{KL}$ distance away from the current distribution. In order to calculate $D_{KL}$ between $p(y; \theta)$ and $p(y; \theta + d)$, where $d \in \mathbb{R}^n$ is a small magnitude “delta” vector, we approximate it using the Fisher Information at $\theta$. Eventually $d$ will be the natural gradient update we will add to $\theta$. To approximate the KL-divergence with Fisher Information, we will start with the Taylor Series expansion of $D_{KL}$ and see that the Fisher Information pops up in the expansion.

Show that $D_{KL}(p_\theta || p_{\theta+d}) \approx \frac{1}{2} d^T I(\theta) d$.

**Hint:** Start with the Taylor Series expansion of $D_{KL}(p_\theta || p_{\tilde{\theta}})$ where $\theta$ is a constant and $\tilde{\theta}$ is a variable. Later set $\tilde{\theta} = \theta + d$. Recall that the Taylor Series allows us to approximate a scalar function $f(\tilde{\theta})$ near $\theta$ by:

$$ f(\tilde{\theta}) \approx f(\theta) + (\tilde{\theta} - \theta)^T \nabla_{\theta'} f(\theta')|_{\theta'=\theta} + \frac{1}{2} (\tilde{\theta} - \theta)^T (\nabla_{\theta'}^2 f(\theta')|_{\theta'=\theta}) (\tilde{\theta} - \theta) $$

### Solution 3(d)

We define the function $f(\tilde{\theta})$ as the KL divergence between $p(y; \theta)$ and $p(y; \tilde{\theta})$:
$$
f(\tilde{\theta}) = D_{KL}(p_{\theta} || p_{\tilde{\theta}}) = \mathbb{E}_{y \sim p(y; \theta)} [\log p(y; \theta) - \log p(y; \tilde{\theta})]
$$

We want to approximate $f(\theta + d)$ using a second-order Taylor expansion around $\theta$:
$$
f(\theta + d) \approx f(\theta) + d^T \nabla f(\theta) + \frac{1}{2} d^T \nabla^2 f(\theta) d
$$

**1. Evaluated at $\theta$:**
$$
f(\theta) = D_{KL}(p_{\theta} || p_{\theta}) = 0
$$

**2. First Derivative (Gradient):**
Take the gradient with respect to $\tilde{\theta}$:
$$
\nabla_{\tilde{\theta}} f(\tilde{\theta}) = \nabla_{\tilde{\theta}} \mathbb{E}_{y \sim p(y; \theta)} [\log p(y; \theta) - \log p(y; \tilde{\theta})]
$$
Since $\log p(y; \theta)$ does not depend on $\tilde{\theta}$:
$$
\nabla_{\tilde{\theta}} f(\tilde{\theta}) = -\mathbb{E}_{y \sim p(y; \theta)} [\nabla_{\tilde{\theta}} \log p(y; \tilde{\theta})]
$$
Evaluating at $\tilde{\theta} = \theta$, we use the result from **3(a)** (expected score is 0):
$$
\nabla f(\theta) = -\mathbb{E}_{y \sim p(y; \theta)} [\nabla_{\theta} \log p(y; \theta)] = 0
$$

**3. Second Derivative (Hessian):**
Take the gradient of the first derivative result with respect to $\tilde{\theta}$ again:
$$
\nabla^2_{\tilde{\theta}} f(\tilde{\theta}) = \nabla_{\tilde{\theta}} (-\mathbb{E}_{y \sim p(y; \theta)} [\nabla_{\tilde{\theta}} \log p(y; \tilde{\theta})]) = -\mathbb{E}_{y \sim p(y; \theta)} [\nabla^2_{\tilde{\theta}} \log p(y; \tilde{\theta})]
$$
Evaluating at $\tilde{\theta} = \theta$, we use the result from **3(c)** ($I(\theta) = -\mathbb{E}[\nabla^2 \log p(y;\theta)]$):
$$
\nabla^2 f(\theta) = -\mathbb{E}_{y \sim p(y; \theta)} [\nabla^2_{\theta} \log p(y; \theta)] = I(\theta)
$$

**Conclusion:**
Substituting these terms back into the Taylor expansion:
$$
D_{KL}(p_{\theta} || p_{\theta+d}) \approx 0 + d^T (0) + \frac{1}{2} d^T I(\theta) d = \frac{1}{2} d^T I(\theta) d
$$

Thus, we have shown the approximation.

### **3(e) [8 points] Natural Gradient**

Now we move on to calculating the natural gradient. Recall that we want to maximize the log-likelihood by moving only by a fixed $D_{KL}$ distance from the current position. In the previous sub-question we came up with a way to approximate $D_{KL}$ distance with Fisher Information. Now we will set up the constrained optimization problem that will yield the natural gradient update $d$. Let the log-likelihood objective be $\ell(\theta) = \log p(y; \theta)$. Let the $D_{KL}$ distance we want to move by, be some small positive constant $c$. The natural gradient update $d^*$ is

$$ d^* = \text{arg max}_d \ell(\theta + d) \quad \text{subject to} \quad D_{KL}(p_\theta \| p_{\theta+d}) = c \quad (1) $$

First we note that we can use Taylor approximation on $\ell(\theta + d) \approx \ell(\theta) + d^T \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta}$. Also note that we calculated the Taylor approximation $D_{KL}(p_\theta \| p_{\theta+d})$ in the previous sub-problem. We shall substitute both these approximations into the above constrained optimization problem.

In order to solve this constrained optimization problem, we employ the method of Lagrange multipliers. If you are familiar with Lagrange multipliers, you can proceed directly to solve for $d^*$. If you are not familiar with Lagrange multipliers, here is a simplified introduction.

Consider the following constrained optimization problem:
$$ d^* = \text{arg max}_d f(d) \quad \text{subject to} \quad g(d) = c $$

The function $f$ is the objective function and $g$ is the constraint. We instead optimize the Lagrangian $L(d, \lambda)$, which is defined as
$$ L(d, \lambda) = f(d) - \lambda [g(d) - c] $$
with respect to both $d$ and $\lambda$. Here $\lambda \in \mathbb{R}^+$ is called the Lagrange multiplier. In order to optimize the above, we construct the following system of equations:
$$ \nabla_d L(d, \lambda) = 0 \quad (a) $$
$$ \nabla_\lambda L(d, \lambda) = 0 \quad (b) $$

So we have two equations (a and b above) with two unknowns ($d$ and $\lambda$), which can sometimes be solved analytically (in our case, we can).

The following steps guide you through solving the constrained optimization problem:

1.  Construct the Lagrangian for the constrained optimization problem (1) with the Taylor approximations substituted in for both the objective and the constraint.
2.  Then construct the system of linear equations (like (a) and (b)) from the Lagrangian you obtained.
3.  From (a), come up with an expression for $d$ that involves $\lambda$. At this stage we have already found the “direction” of the natural gradient $d$, since $\lambda$ is only a positive scaling constant. For most practical purposes, the solution we obtain here is sufficient. This is because we almost always include a learning rate hyperparameter in our optimization algorithms, or perform some kind of a line search for algorithmic stability. This can make the exact calculation of $\lambda$ less critical. Let’s call this expression $\tilde{d}$ (involving $\lambda$) as the unscaled natural gradient. Clearly state what is $\tilde{d}$ as a function of $\lambda$. The remaining steps are to figure out the value of the scaling constant $\lambda$ along the direction of $d$, for completeness.
4.  Plug that expression for $d$ into (b). Now we have an equation that has $\lambda$ but not $d$. Come up with an expression for $\lambda$ that does not include $d$.
5.  Plug that expression for $\lambda$ (without $d$) back into (a). Now we have an equation that has $d$ but not $\lambda$. Come up with an expression for $d$ that does not include $\lambda$.

The expression of $d$ obtained this way will be the desired natural gradient update $d^*$. **Clearly state and highlight your final expression for $d^*$. This expression cannot include $\lambda$.**

### Solution 3(e)

**1. Constructing the Lagrangian**

We want to maximize the linearized log-likelihood:
$$ l(\theta + d) \approx l(\theta) + d^T \nabla_\theta l(\theta) $$

Subject to the constraint on the approximated KL divergence:
$$ \frac{1}{2} d^T I(\theta) d = c $$

The Lagrangian is:
$$ \mathcal{L}(d, \lambda) = (l(\theta) + d^T \nabla_\theta l(\theta)) - \lambda \left( \frac{1}{2} d^T I(\theta) d - c \right) $$

**2. System of Equations**

Taking the gradient with respect to $d$:
$$ \nabla_d \mathcal{L} = \nabla_\theta l(\theta) - \lambda I(\theta) d = 0 \quad (a) $$

The constraint equation:
$$ \frac{1}{2} d^T I(\theta) d = c \quad (b) $$

**3. Finding $\tilde{d}$ as a function of $\lambda$**

From (a), we solve for $d$:
$$ \lambda I(\theta) d = \nabla_\theta l(\theta) $$
$$ d = \frac{1}{\lambda} I(\theta)^{-1} \nabla_\theta l(\theta) $$

Letting $\tilde{d}$ denote this unscaled direction:
$$ \tilde{d} = I(\theta)^{-1} \nabla_\theta l(\theta) $$
So $d = \frac{1}{\lambda} \tilde{d}$.

**4. Finding $\lambda$**

Substitute $d = \frac{1}{\lambda} \tilde{d}$ into (b):
$$ \frac{1}{2} \left( \frac{1}{\lambda} \tilde{d} \right)^T I(\theta) \left( \frac{1}{\lambda} \tilde{d} \right) = c $$
$$ \frac{1}{2\lambda^2} \tilde{d}^T I(\theta) \tilde{d} = c $$
$$ \lambda^2 = \frac{1}{2c} \tilde{d}^T I(\theta) \tilde{d} $$
$$ \lambda = \sqrt{\frac{\tilde{d}^T I(\theta) \tilde{d}}{2c}} $$

**5. Final Expression for $d^*$**

Substitute $\lambda$ back into the expression for $d$:
$$ d^* = \frac{1}{\sqrt{\frac{\tilde{d}^T I(\theta) \tilde{d}}{2c}}} \tilde{d} $$

Simplifying:
$$ d^* = \frac{\sqrt{2c}}{\sqrt{\tilde{d}^T I(\theta) \tilde{d}}} \tilde{d} $$

Substituting back $\tilde{d} = I(\theta)^{-1} \nabla_\theta l(\theta)$:
$$ d^* = \frac{\sqrt{2c}}{\sqrt{(I(\theta)^{-1} \nabla_\theta l(\theta))^T I(\theta) (I(\theta)^{-1} \nabla_\theta l(\theta))}} I(\theta)^{-1} \nabla_\theta l(\theta) $$

Note that the denominator is a scalar normalization factor. Thus, the direction is indeed given by the natural gradient $I^{-1} \nabla l$.

### **3(f) [2 points] Relation to Newton’s Method**

After going through all these steps to calculate the natural gradient, you might wonder if this is something used in practice. We will now see that the familiar Newton’s method that we studied earlier, when applied to Generalized Linear Models, is equivalent to natural gradient on Generalized Linear Models. While the two methods (Netwon’s and natural gradient) agree on GLMs, in general they need not be equivalent.

Show that the direction of update of Newton’s method, and the direction of natural gradient, are exactly the same for Generalized Linear Models. You may want to recall and cite the results you derived in problem set 1 question 4 (Convexity of GLMs). For the natural gradient, it is sufficient to use $\tilde{d},$ the unscaled natural gradient.

---

### **Solution**

**1. Definitions of the Updates**
*   **Natural Gradient:** The unscaled natural gradient direction is defined as the product of the inverse Fisher Information matrix and the gradient of the log-likelihood (score function):
    $$ \tilde{d}_{natural} = I(\theta)^{-1} \nabla_\theta \ell(\theta) $$
    where $I(\theta) = E_{y \sim p(y;\theta)}[-\nabla^2_\theta \ell(\theta)]$ (from part 3(c)).

*   **Newton's Method:** Newton's method updates parameters using the inverse Hessian of the log-likelihood:
    $$ d_{newton} = -(\nabla^2_\theta \ell(\theta))^{-1} \nabla_\theta \ell(\theta) $$

**2. Property of GLMs**
Recall from Problem Set 1, Question 4 (Convexity of GLMs) that for a Generalized Linear Model in canonical form, the log-likelihood is concave. Specifically, the Hessian of the log-likelihood ($\nabla^2_\theta \ell(\theta)$) depends only on the input features $x$ and the model parameters $\theta$, but **not** on the random variable/labels $y$.

For a GLM, the Hessian takes the form:
$$ \nabla^2_\theta \ell(\theta) = -X^T W X $$
where $X$ is the design matrix and $W$ is a diagonal matrix of variances which depends only on $\theta$ and $x$, not $y$.

**3. Equivalence**
Because the Hessian $\nabla^2_\theta \ell(\theta)$ is constant with respect to $y$, its expected value is simply itself:
$$ E_{y}[-\nabla^2_\theta \ell(\theta)] = -\nabla^2_\theta \ell(\theta) $$

Since the Fisher Information $I(\theta)$ is defined as the expected negative Hessian:
$$ I(\theta) = -\nabla^2_\theta \ell(\theta) $$

Substituting this into the Natural Gradient equation:
$$ \tilde{d}_{natural} = (-\nabla^2_\theta \ell(\theta))^{-1} \nabla_\theta \ell(\theta) $$
$$ \tilde{d}_{natural} = -(\nabla^2_\theta \ell(\theta))^{-1} \nabla_\theta \ell(\theta) $$

This is exactly the update direction for Newton's method. Thus, for GLMs, Natural Gradient descent is equivalent to Newton's method.
